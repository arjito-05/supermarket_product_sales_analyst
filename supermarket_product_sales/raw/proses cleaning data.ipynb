# melakukan data cleaning pada dataset supermarket product sales

melakukan load dataset yang ingin di cleaning

```python
import pandas as pd
import pyarrow

df = pd.read_csv('supermarket_product_sales.csv')
print(df.info)
```

```python
df.info()
```

melakukan perubahan data type pada order date dan ship date dari 'object' menjadi 'datetime64'

```python
df['Order Date'] = pd.to_datetime(df['Order Date'], dayfirst = True)
df['Ship Date'] = pd.to_datetime(df['Ship Date'], dayfirst = True)
df.info()
```

mencari baris yang kosong

```python
df.isnull().sum()
```

terdapat baris yang kosong pada kolom Postal Code namun karena kolom tersebut bukan variabel utama jadi saya isi dengan '0' agar mencegah error atau sebagai penanda bahwa baris tersebut kosong

```python
df['Postal Code'] = df['Postal Code'].fillna(0)
df.isna().sum()
```

baik untuk sekarang dataset tersebut sudah bersih dan siap untuk menjadi bahan analisis

```python
df.to_parquet('supermarket_product_sales_clean.parquet')
```
