# supermarket product sales data analyst

melakukan load dataset supermarket product sales yang sudah clean

```python
import pandas as pd
import pyarrow

df = pd.read_parquet('supermarket_product_sales_clean.parquet')
print(df.info)
```

```python
df.info()
```

 ## melakukan analisis penjualan

```python
df['Sales'].describe()
```

menghitung total sales

```python
Total_sales = df['Sales'].sum()
print(f"Total_sales : ${Total_sales:,.2f}")
```

menghitung rata-rata sales

```python
avarage_sales = df['Sales'].mean()
print(f"avarage_sales : ${avarage_sales:,.2f}")
```

analisis tren penjualan

```python
df['Year'] = df['Order Date'].dt.year
df['Month'] = df['Order Date'].dt.month
```

total penjualan pertahun

```python
sales_by_year = (df.groupby('Year')['Sales'].sum())
sales_by_year
```

Melakukan analisis kategori Produk

Kategori produk dengan penjualan terbesar

```python
sales_by_category = (
    df.groupby('Category')['Sales'].sum().sort_values(ascending = False)
)
sales_by_category
```

Sub-Kategori dengan penjualan terbesar

```python
sales_by_sub_category = (
    df.groupby('Sub-Category')['Sales'].sum().sort_values(ascending = False)
)
sales_by_sub_category
```

melakukan analisis wilayah

wilyah dengan penjualan terbesar

```python
sales_by_region = (
    df.groupby('Region')['Sales'].sum().sort_values(ascending = False)
)
sales_by_region
```

kota dengan penjualan terbesar

```python
sales_by_city = (
    df.groupby('City')['Sales'].sum().sort_values(ascending = False)
)
sales_by_city.head(10)
```

Segment pelanggan dengan kontribusi penjualan terbesar

```python
sales_by_segment = (
    df.groupby('Segment')['Sales'].sum().sort_values(ascending = False)
)
sales_by_segment
```

produk yang menghasilkan penjualan terbesar

```python
sales_by_product_name = (
    df.groupby('Product Name')['Sales'].sum().sort_values(ascending = False)
)
sales_by_product_name.head(10)
```

## melakukan analisis shipping mode

metode pengiriman paling sering digunakan

```python
shipping_mode = (
    df['Ship Mode'].value_counts()
)
shipping_mode
```
